Tensor from Numpy:
both numpy and tensor share same memory

```python
import numpy as np
a=np.array([[1,2,3],[4,5,6]])
b=torch.from_numpy(a)  #converting numpy array to torch tensor
print(b)
```

Matrix calculator(small assignment)

```python
mat1=torch.tensor([[1,2,3],[4,5,6]])
mat2=torch.tensor([[3,2,1],[6,5,4],[9,8,7]])
result=torch.matmul(mat1, mat2)
print("result matrix:", result) 
print("shape of the result is:", result.shape)  #row and column of result matrix
print(torch.mm(mat1, mat2))  #another way of doing matrix multiplication
```
